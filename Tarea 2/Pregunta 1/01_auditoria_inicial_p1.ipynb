### 1.4 Análisis de datos faltantes

Se analiza la completitud de las variables seleccionadas para la Pregunta
de Negocio 1. Para cada variable se calcula la cantidad de registros con
información, la cantidad de valores faltantes y su participación
porcentual.

En esta etapa no se realiza imputación ni eliminación de registros.

```python
# Calculamos cantidad y porcentaje de valores faltantes

faltantes_p1 = pd.DataFrame({
    "no_nulos": df[variables_p1].notna().sum(),
    "faltantes": df[variables_p1].isna().sum()
})

faltantes_p1["porcentaje_faltantes"] = (
    faltantes_p1["faltantes"] / len(df) * 100
)

faltantes_p1 = faltantes_p1.sort_values(
    "porcentaje_faltantes",
    ascending=False
)

faltantes_p1
```

#### Hallazgo

La revisión de completitud evidencia dos patrones principales de datos
faltantes.

En primer lugar, `fecha_de_firma` presenta 8.128 registros faltantes,
equivalentes al 31,74% del conjunto de datos, siendo la variable
seleccionada con mayor proporción de información ausente.

En segundo lugar, `proveedor_adjudicado`, `valor_del_contrato`,
`modalidad_de_contratacion`, `tipo_de_contrato`, `es_grupo` y
`es_pyme` presentan cada una 4.887 registros faltantes, equivalentes al
19,09% de la base.

Por el contrario, `id_contrato`, `referencia_del_contrato` y
`estado_contrato` presentan información completa para los 25.605
registros.

Debido a que varias variables presentan exactamente la misma cantidad
de datos faltantes, se requiere verificar si corresponden a los mismos
registros antes de definir cualquier estrategia de tratamiento.

### 1.5 Análisis del patrón de datos faltantes

Se observa que varias variables relevantes presentan la misma cantidad
de datos faltantes. Por esta razón se verifica si la ausencia de
información corresponde a los mismos registros.

Esta revisión permitirá determinar si existe un patrón estructural de
información incompleta dentro de la fuente de datos.

```python
# Variables que presentan potencialmente el mismo patrón de faltantes

variables_patron = [
    "proveedor_adjudicado",
    "valor_del_contrato",
    "modalidad_de_contratacion",
    "tipo_de_contrato",
    "es_grupo",
    "es_pyme"
]

# Revisamos cuántos registros tienen todas estas variables faltantes simultáneamente

patron_faltantes = df[variables_patron].isna().all(axis=1)

print(
    "Registros con todas las variables simultáneamente faltantes:",
    patron_faltantes.sum()
)

print(
    "Porcentaje:",
    round(patron_faltantes.mean() * 100, 2),
    "%"
)
```

```python
# Revisamos si los registros sin proveedor también carecen de las demás variables clave

sin_proveedor = df[df["proveedor_adjudicado"].isna()]

sin_proveedor[variables_patron].notna().sum()
```

```python
# Revisamos el estado contractual de los registros con el patrón de faltantes

df.loc[
    patron_faltantes,
    "estado_contrato"
].value_counts(dropna=False)
```

```python
# Observamos una muestra de estos registros

df.loc[
    patron_faltantes,
    [
        "id_contrato",
        "referencia_del_contrato",
        "estado_contrato",
        "fecha_de_firma",
        "proveedor_adjudicado",
        "valor_del_contrato"
    ]
].head(15)
```

#### Hallazgo

El análisis conjunto de los valores faltantes confirma la existencia de
un patrón estructural de información incompleta.

Los 4.887 registros que no contienen información en
`proveedor_adjudicado` carecen simultáneamente de información en
`valor_del_contrato`, `modalidad_de_contratacion`,
`tipo_de_contrato`, `es_grupo` y `es_pyme`. Estos registros
representan aproximadamente el 19,09% del conjunto de datos.

Por lo tanto, los faltantes encontrados en estas seis variables no
deben analizarse como problemas independientes, sino como un mismo
grupo de registros con información contractual incompleta.

Adicionalmente, la ausencia de `fecha_de_firma` presenta un patrón más
amplio: existen 8.128 registros sin fecha de firma, por lo que 3.241
casos adicionales carecen de fecha aun cuando pueden disponer de
información en las demás variables principales.

En esta etapa no se eliminan estos registros. En las siguientes
secciones se analizarán sus estados contractuales y demás
características para determinar si corresponde excluirlos de la
población analítica o tratarlos de manera diferenciada.

### 1.14 Evaluación preliminar de la población analítica

Antes de definir el conjunto definitivo para el análisis, se compara la
cantidad de registros disponibles bajo diferentes criterios relacionados
con la existencia de proveedor, valor contractual positivo y fecha de
firma.

Esta evaluación es diagnóstica y no implica aún la eliminación de
registros.

```python
# Condiciones preliminares

condicion_proveedor = (df["proveedor_adjudicado"].notna())

condicion_valor_positivo = (df["valor_del_contrato"] > 0)

condicion_fecha_firma = (fecha_firma_dt.notna())

print("Registros totales:",len(df))

print("Con proveedor:",condicion_proveedor.sum())

print("Con valor positivo:",condicion_valor_positivo.sum())

print("Con fecha de firma:",condicion_fecha_firma.sum())

print("Con proveedor + valor positivo:",(condicion_proveedor& condicion_valor_positivo).sum())

print("Con proveedor + valor positivo + fecha de firma:",(condicion_proveedor& condicion_valor_positivo& condicion_fecha_firma).sum())
```
