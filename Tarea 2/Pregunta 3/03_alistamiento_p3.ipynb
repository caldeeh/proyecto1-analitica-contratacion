## 3. Revisión de variables seleccionadas

A partir de las variables disponibles se define un conjunto de variables de trabajo orientado a las dimensiones financiera, contractual, temporal, institucional y de proveedor.

Antes de construir la base analítica se revisan sus tipos de datos y valores faltantes, con el fin de asegurar que las variables seleccionadas puedan utilizarse adecuadamente en los análisis posteriores.

```python
# Variables seleccionadas para el análisis
variables_seleccionadas = [
    "id_contrato",
    "referencia_del_contrato",
    "nombre_entidad",
    "departamento",
    "ciudad",
    "sector",
    "rama",
    "estado_contrato",
    "tipo_de_contrato",
    "modalidad_de_contratacion",
    "descripcion_del_proceso",
    "fecha_de_firma",
    "fecha_de_inicio_del_contrato",
    "fecha_de_fin_del_contrato",
    "duracion_del_contrato",
    "dias_adicionados",
    "proveedor_adjudicado",
    "es_grupo",
    "es_pyme",
    "valor_del_contrato",
    "valor_facturado",
    "valor_pagado",
    "valor_pendiente_de_pago",
    "valor_pendiente_de_ejecucion",
    "saldo_cdp",
    "saldo_vigencia",
    "el_contrato_puede_ser_prorrogado"
]

print(f"Variables seleccionadas: {len(variables_seleccionadas)}")
```

## 4. Construcción de la base analítica

Con base en las variables seleccionadas se construye una base analítica específica para la Pregunta 3.

Esta base conserva únicamente la información necesaria para caracterizar los contratos y desarrollar los análisis posteriores, manteniendo las variables financieras, contractuales, temporales, institucionales y de proveedor identificadas previamente.

Los valores faltantes presentes en algunas variables temporales se conservarán, dado que corresponden a información no disponible en determinados registros y no constituyen, por sí mismos, una razón para excluir los contratos del análisis.

```python
# Construcción de la base analítica
df_analitico = df[variables_seleccionadas].copy()

print("Base analítica construida correctamente.")
print(f"Filas: {df_analitico.shape[0]}")
print(f"Columnas: {df_analitico.shape[1]}")
```

```python
df_analitico.info()
```

## 5. Construcción de indicadores financieros

Para facilitar la interpretación del comportamiento financiero de los contratos se incorporan indicadores relativos de ejecución y pago.

Estos indicadores permiten comparar contratos de diferentes magnitudes y serán utilizados posteriormente para la exploración de los datos y la construcción de visualizaciones.

El porcentaje ejecutado relaciona el valor facturado con el valor contractual, mientras que el porcentaje pagado relaciona el valor pagado con el valor contractual.

```python
# Construcción de indicadores financieros relativos

df_analitico["porcentaje_ejecutado"] = np.where(
    df_analitico["valor_del_contrato"] > 0,
    (df_analitico["valor_facturado"] / df_analitico["valor_del_contrato"]) * 100,
    np.nan
)

df_analitico["porcentaje_pagado"] = np.where(
    df_analitico["valor_del_contrato"] > 0,
    (df_analitico["valor_pagado"] / df_analitico["valor_del_contrato"]) * 100,
    np.nan
)

print("Indicadores financieros creados correctamente.")
```

```python
df_analitico[
    ["valor_del_contrato", "porcentaje_ejecutado", "porcentaje_pagado"]
].describe()
```

## 6. Indicador de extensión contractual

Se incorpora una variable categórica que identifica si el contrato presenta días adicionados a su duración.

Esta clasificación permite diferenciar los contratos que no presentan extensiones de aquellos que han tenido modificaciones en su plazo, facilitando su análisis conjunto con las variables financieras y contractuales.

```python
# Clasificación de contratos según días adicionados

df_analitico["tiene_extension"] = np.where(
    df_analitico["dias_adicionados"] > 0,
    "Sí",
    "No"
)

print("Indicador de extensión contractual creado correctamente.")
print(df_analitico["tiene_extension"].value_counts())
```

## 7. Identificación de contratos prioritarios

Se construye un indicador de priorización utilizando simultáneamente dos dimensiones identificadas durante el análisis: el valor pendiente de ejecución y los días adicionados al contrato.

Como umbrales se utilizan los percentiles 90 (P90) de ambas variables. Un contrato se clasifica como prioritario cuando supera simultáneamente el P90 del valor pendiente de ejecución y el P90 de los días adicionados.

Este indicador permitirá orientar posteriormente la exploración y visualización hacia los contratos que presentan simultáneamente una mayor exposición financiera pendiente y una extensión contractual superior a la observada en la mayoría de los registros.

```python
# Umbrales de priorización
p90_pendiente_ejecucion = df_analitico[
    "valor_pendiente_de_ejecucion"
].quantile(0.90)

p90_dias_adicionados = df_analitico[
    "dias_adicionados"
].quantile(0.90)

# Clasificación de contratos prioritarios
df_analitico["contrato_prioritario"] = np.where(
    (df_analitico["valor_pendiente_de_ejecucion"] > p90_pendiente_ejecucion) &
    (df_analitico["dias_adicionados"] > p90_dias_adicionados),
    "Sí",
    "No"
)

print(f"P90 pendiente de ejecución: {p90_pendiente_ejecucion:,.1f}")
print(f"P90 días adicionados: {p90_dias_adicionados:,.1f}")
print(
    "Contratos prioritarios:",
    (df_analitico["contrato_prioritario"] == "Sí").sum()
)
```

## 8. Verificación final de la base analítica

Se realiza una verificación final de la estructura de la base analítica y de los indicadores construidos, antes de utilizarla como insumo para la etapa de exploración y análisis de datos.
